## 15
Leia a altura e o peso de uma pessoa e calcule o IMC, informando a classificação.

```
peso = float(input("Digite seu peso: "))
altura = float(input("Digite a sua altura: "))

imc = peso / (altura ** 2)
print("Seu IMC é de: ",imc)

if imc >= 18.5 and imc <= 24.9:
  print("Peso normal")
elif imc >= 25.0 and imc <= 29.9:
  print("Sobrepeso")
elif imc >= 30.0 and imc <= 34.9:
  print("Obesidade garu 1")
elif imc >= 35.0 and imc <= 39.9:
  print("Obesidade garu 2")
else:
  print("Obesidade garu 3")
```

## 17
Leia 10 números e informe quantos são pares e quantos são ímpares.

```
pares = 0
impares = 0

for i in range(1, 11):
    n = int(input("Digite um número: "))

    if n % 2 == 0:
        pares += 1
    else:
        impares += 1

print("Quantidade de números pares:", pares)
print("Quantidade de números ímpares:", impares)
```

## 18
Leia um número N e gere os N primeiros termos da sequência de Fibonacci.

```
n = int(input("Digite a quantidade de termos: "))

a = 0
b = 1

for i in range(n):
    print(a, end=" ")
    proximo = a + b
    a = b
    b = proximo
```

## 19
Crie um sistema de login com usuário e senha. Permita até 3 tentativas antes de bloquear o acesso.

```
for tentativa in range(3):
    usuario = input("Usuário: ")
    senha = input("Senha: ")

    if usuario == "admin" and senha == "1234":
        print("Login realizado com sucesso!")
        break
    else:
        restantes = 2 - tentativa
        print("Dados incorretos")

        if restantes > 0:
            print(f"Você só tem mais {restantes} tentativa(s)")
else:
    print("Acesso bloqueado!")
```

## 20
Crie um menu interativo com as opções:
1 - Somar
2 - Subtrair
3 - Multiplicar
4 - Dividir
0 - Sair

O programa deve continuar executando até o usuário escolher sair.

```

while True:
    print("==================================")
    print("1 - Para SOMAR")
    print("2 - Para SUBTRAIR")
    print("3 - Para MULTPLICAR")
    print("4 - Para DIVIIDR")
    print("0 - Para SAIR")
    print("==================================")
    opcao = int(input("Digite uma opção: "))
    match opcao:
      case 1:
        soma()
      case 2:
        sub()
      case 3:
        multiplicacao()
      case 4:
        divisao()
      case 0:
        print("Encerrando sistema")
        break
      case _:
        print("Opção invalidad")


def soma():
    n1= int(input("Digite um numero: "))
    n2 = int(input("Digite outro numero: "))

    resultado = n1 + n2
    print("A soma dos dois numeros é de ",resultado)

def sub():
    n1= int(input("Digite um numero: "))
    n2= int(input("Digite outro numero: "))

    resultado = n1 - n2
    print("A subtração dos dois numeros é de ",resultado)

def multiplicacao():
    n1= int(input("Digite um numero: "))
    n2= int(input("Digite outro numero: "))

    resultado = n1 * n2
    print("A multiplicação dos dois numeros é de ",resultado)

def divisao():

    n1= int(input("Digite um numero: "))
    n2= int(input("Digite outro numero: "))

    if n2 == 0:
        print("Erro: divisão por zero não é permitida.")
    else:
        resultado = n1 / n2
        print("Resultado da divisão:", resultado)
```
